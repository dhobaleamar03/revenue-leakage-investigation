We identified **673 negative values** in the `Demand Forecast` column. Since a demand forecast cannot logically be negative, these values are treated as invalid and corrected during the data-cleaning stage by replacing them with **0**.

This ensures the cleaned dataset contains only valid non-negative demand forecast values.

## DATA CLEANING

```python
# Clean negative Demand Forecast values
negative_forecast = (df["Demand Forecast"] < 0).sum()
print("Negative Demand Forecast values found:", negative_forecast)
```

```python
# Replace all negative Demand Forecast values with 0
df.loc[df["Demand Forecast"] < 0, "Demand Forecast"] = 0
```

```python
# Verify that no negative Demand Forecast values remain
remaining_negative = (df["Demand Forecast"] < 0).sum()
print("Negative values remaining after cleaning:", remaining_negative)
```

```python
df.to_csv("cleaned_inventory.csv", index=False)

print("Cleaned dataset saved")
```
